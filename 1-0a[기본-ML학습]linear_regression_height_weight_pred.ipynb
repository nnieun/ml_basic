- 몸무게 : lb(파운트) -> kg, 1lb = 0.4563kg 

#### 방법

```python
# read weight-height dataset
data = "datas_ml/weight-height.csv"
wh_df = pd.read_csv(data)

# 키 변경하기 ( inch -> cm)
wh_df["Height"] = wh_df["Height"]* 2.54

# 몸무게 변경하기 ( lb -> kg )
wh_df["Weight"] = wh_df["Weight"] * 0.4563

wh_df.head()
```

# 1. 키에 따른 몸무게 예측
## 학습, 테스트 데이터 준비
- Feature(x)와 label(y) 데이터 분리 

```python
x = wh_df["Height"] # 입력데이터
y = wh_df["Weight"] # 정답데이터

# 일반적으로 소문자 씀 
```

```python
y[:5]
```

## 80%는 트레이닝 데이터, 20%는 테스트 데이터로 나누기
- random_seed 값 주기
- stratify=y 설정은
  - target이 classification인 경우
  - feature가 2개이상 있을 때 사용

```python
# 학습, 테스트 데이터셋으로 분할
x_train, x_test, y_train, y_test = train_test_split(
    x, 
    y, 
    test_size=0.2, # 분할 비율 8:2 로 쪼갠다.
    random_state=10 # 데이터 분할 재현 4개로 쪼개질 때 데이터의 구성을 재현한다.
)

# 구조가 정형화 되어있기때문에 한번에 쪼개는거임 

```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

## 회귀 모델 학습하기
- 선형회귀(Linear Regression

```python
# 회귀 모델 알고리즘 객체 생성 
model_lr = LinearRegression()
```

```python
# x는 2차원이어야 함.
x_train_2d = x_train.values.reshape(-1,1)
# x_train.values 넘파이 객체 
```

```python
x_train_2d.shape
# (8000, 1)
```
